### Step 3 Budget exploration
Budget is strongly right-skewed. The raw and `log10` histograms show how a log transformation compresses its range. These graphs are retained for interpretation; `log_budget = log10(budget)` is selected as a numeric predictor; raw budget remains excluded.

### Step 4 Genre flags
A movie can have several genres in one `|` separated text field, which models cannot use directly. Each genre becomes its own 0/1 column, so a movie that is both Action and Comedy gets a 1 in both. The bar chart shows how often each genre appears. "TV Movie" appears only 3 times, which is too few to learn from, so it gets no column. `n_genres` counts how many genres a movie lists.

```python
#  Genre flags
# genres is a "|" list, and a movie can have several, so each genre gets its own 0/1 column
genre_counts = movies["genres"].str.split("|").explode().value_counts()
print(genre_counts)

fig, ax = plt.subplots(figsize=(8, 5))
ax.barh(genre_counts.index[::-1], genre_counts.values[::-1], color=BLUE)
ax.set_xlabel("Number of movies")
ax.set_title("Movies per genre (a movie can have several)")
plt.tight_layout()
plt.show()
```

```python
# "TV Movie" has only 3 movies, too few to learn from, so it gets no flag
genre_list = []
for genre in genre_counts.index:
    if genre != "TV Movie":
        genre_list.append(genre)

genre_columns = []
for genre in genre_list:
    column = "genre_" + genre.lower().replace(" ", "_")          # e.g. "Science Fiction" -> genre_science_fiction
    movies[column] = movies["genres"].str.split("|").apply(lambda names: int(genre in names))
    genre_columns.append(column)

# Number of genres listed for each movie
movies["n_genres"] = movies["genres"].str.split("|").apply(len)

print("Genre flags created:", len(genre_columns))
movies[["title_tmdb", "genres", "n_genres"] + genre_columns[:5]].head()
```

### Step 6 Director experience from IMDb
A director's track record is known before release and may signal the chance of success. Using the IMDb director IDs, this step counts how many feature films the movie's directors made before this movie, and how many years have passed since their first one. Only films from earlier IMDb years count, and IMDb years are compared only with IMDb years. Because IMDb covers each director's full career, these counts are not limited by our data starting in 1990. Directors with no earlier film are flagged as first-time directors.

```python
# Director career length from IMDb (all of a director's earlier films, not just those in our data)
# Rule: only films from EARLIER IMDb years count (IMDb years are compared only with IMDb years)

# Build a lookup: director ID -> list of (year, imdb_id) for every IMDb movie they directed
director_career = {}
for director_id, film_id, year in zip(imdb["nconst"], imdb["imdb_id"], imdb["startYear"]):
    if director_id not in director_career:
        director_career[director_id] = []
    director_career[director_id].append((year, film_id))

def director_history_imdb(director_ids, this_year):
    """Count distinct earlier films by any of the movie's directors, and years since their first film."""
    earlier_films = set()
    first_year = None
    for director_id in director_ids.split("|"):
        for year, film_id in director_career.get(director_id, []):
            if year < this_year:
                earlier_films.add(film_id)
                if first_year is None or year < first_year:
                    first_year = year
    if first_year is None:
        return 0, 0
    return len(earlier_films), this_year - first_year

prior_films = []
years_active = []
for director_ids, this_year in zip(movies["imdb_director_ids"], movies["startYear"]):
    count, years = director_history_imdb(director_ids, this_year)
    prior_films.append(count)
    years_active.append(years)

movies["dir_prior_films"] = prior_films          # earlier feature films directed (IMDb)
movies["dir_years_active"] = years_active        # years since the director's first feature film
movies["dir_first_film"] = (movies["dir_prior_films"] == 0).astype(int)   # first-time director flag

print(movies[["dir_prior_films", "dir_years_active"]].describe().round(1))
print("First-time directors:", movies["dir_first_film"].sum())
```

### Step 7 Director track record in our data
Count distinct earlier-year films in the cleaned financial cohort, pooling co-directors without counting a film twice. The raw success fraction is missing when no earlier record exists; keep a separate no-record flag. A later step selects a smoothed rate instead, using a five-film prior and earlier-year global outcomes. Earlier-year ordering does not prove final box office was reported before the target film's release.

```python
#  Director track record inside our data (box-office outcomes of their earlier films)
# Rule: only films from EARLIER TMDB release years count, as a conservative ordering rule; historical reporting availability is not verified
director_outcomes = {}
for director_ids, film_id, year, success in zip(movies["imdb_director_ids"], movies["tmdb_id"], movies["release_year"], movies["success"]):
    for director_id in director_ids.split("|"):
        if director_id not in director_outcomes:
            director_outcomes[director_id] = []
        director_outcomes[director_id].append((year, film_id, success))

def director_track_record(director_ids, this_year):
    """Number of earlier films with known outcomes, and their success rate (None if there are none)."""
    earlier = {}
    for director_id in director_ids.split("|"):
        for year, film_id, success in director_outcomes.get(director_id, []):
            if year < this_year:
                earlier[film_id] = success          # dictionary keeps each film once for co-directors
    if len(earlier) == 0:
        return 0, np.nan
    return len(earlier), sum(earlier.values()) / len(earlier)

known_films = []
success_rates = []
for director_ids, this_year in zip(movies["imdb_director_ids"], movies["release_year"]):
    count, rate = director_track_record(director_ids, this_year)
    known_films.append(count)
    success_rates.append(rate)

movies["dir_prior_known_films"] = known_films         # earlier films in our data with a known outcome
movies["dir_prior_success_rate"] = success_rates      # missing when there is no earlier record
movies["dir_no_track_record"] = (movies["dir_prior_known_films"] == 0).astype(int)

print("Directors with no earlier record in our data:", movies["dir_no_track_record"].sum())
print(movies["dir_prior_success_rate"].describe().round(3))
```

The table below checks the history features on the movies with the longest records. The counts and rates look reasonable. Anthology films with many directors, such as "Paris, je t'aime", combine the records of all their directors.

```python
# Look at a few well-known directors to check the history features make sense
director_columns = ["title_tmdb", "release_year", "imdb_directors", "dir_prior_films", "dir_years_active",
                    "dir_prior_known_films", "dir_prior_success_rate"]
movies.sort_values("dir_prior_known_films", ascending=False)[director_columns].head()
```

### Step 8 Lead actor experience: original five-year count
Use the first-listed TMDB actor as the lead. Count matched TMDB movies from the five release years before the current movie in which that actor appears in the top-five cast. Name matching can confuse people with the same name, and the sample does not contain every film.

This step creates the original `lead_actor_films_5yr` count. Because the TMDB sample starts in 1990, early five-year windows are incomplete. **Step 11** adds eligible IMDb credits from 1985–1990 that are absent from the matched TMDB reference sample, producing `lead_actor_films_5yr_backfilled`, then applies `log1p` for the final predictor. Missing or unreliable backfill matches are flagged. The supplement improves coverage but does not provide a complete filmography.

For a 1990 release, the window is 1985–1989; only earlier years count. The supplement is added once in Step 11, not in this original count.

```python
# Original TMDB lead-actor count in the previous five years
# TMDB gives cast names only (no IDs), so the first-billed actor is matched by name.
# Counts films from the 5 years before this movie's year, in all 13,276 matched movies,
# where that actor is in the top 5 billed cast. The sample starts in 1990.
# Step 11 adds the IMDb supplement once, then applies log1p; this remains the original count.
HISTORY_YEARS = 5

actor_years = {}
for cast, year in zip(all_movies["top_cast"].fillna(""), all_movies["release_year"]):
    for name in cast.split("|"):
        if name != "":
            if name not in actor_years:
                actor_years[name] = []
            actor_years[name].append(year)

lead_recent = []
for cast, this_year in zip(movies["top_cast"].fillna(""), movies["release_year"]):
    lead = cast.split("|")[0]
    count = 0
    for year in actor_years.get(lead, []):
        if this_year - HISTORY_YEARS <= year < this_year:
            count += 1
    lead_recent.append(count)

movies["lead_actor_films_5yr"] = lead_recent
print(movies["lead_actor_films_5yr"].describe().round(1))
```

### Step 9 Production company activity in the previous 5 years
Larger and more active studios may have bigger marketing and distribution reach. Instead of hand-picking a list of major studios, this step counts how many films each of the movie's production companies released in the previous 5 years and keeps the largest count. The same 5-year window keeps the feature comparable across years.

```python
#  Production company activity in the previous 5 years
# The largest number of films, from the 5 years before this movie's year, made by any of the
# movie's production companies. This measures studio size and activity without hand-picking
# a list of "major" studios.
company_years = {}
for companies, year in zip(all_movies["production_companies"].fillna(""), all_movies["release_year"]):
    for company in companies.split("|"):
        if company != "":
            if company not in company_years:
                company_years[company] = []
            company_years[company].append(year)

company_recent = []
for companies, this_year in zip(movies["production_companies"].fillna(""), movies["release_year"]):
    best = 0
    for company in companies.split("|"):
        count = 0
        for year in company_years.get(company, []):
            if this_year - HISTORY_YEARS <= year < this_year:
                count += 1
        if count > best:
            best = count
    company_recent.append(best)

movies["company_films_5yr"] = company_recent
print(movies["company_films_5yr"].describe().round(1))
```

### Step 10 Earlier films in the same franchise
`is_sequel` from the cleaning step says whether an earlier film in the collection exists. This step adds how many earlier films the franchise has, since a fourth installment may behave differently from a second. It uses TMDB release dates on both sides, and movies outside a collection get 0.

```python
# Number of earlier films in the same franchise
# Uses TMDB release dates on both sides, like is_sequel; 0 for movies not in a collection
collection_dates = {}
for collection_id, date in zip(all_movies["collection_id"], all_movies["release_date"]):
    if pd.notna(collection_id):
        if collection_id not in collection_dates:
            collection_dates[collection_id] = []
        collection_dates[collection_id].append(date)

franchise_prior = []
for collection_id, this_date in zip(movies["collection_id"], movies["release_date"]):
    count = 0
    if pd.notna(collection_id):
        for date in collection_dates.get(collection_id, []):
            if date < this_date:
                count += 1
    franchise_prior.append(count)

movies["franchise_prior_films"] = franchise_prior
print(movies["franchise_prior_films"].value_counts().sort_index().head(8))
movies.loc[movies["franchise_prior_films"] > 0, ["title_tmdb", "release_year", "collection_name", "is_sequel", "franchise_prior_films"]].head()
```

The table below compares history features across release periods. These are the original counts before the actor supplement and transformations. Early actor/company windows are incomplete because the TMDB reference sample starts in 1990. The later 1985–1990 supplement improves some actor windows; company coverage remains unchanged. A fixed five-year window does not itself guarantee equal coverage.

```python
# Check that history features do not simply grow with the release year
# Our data starts in 1990, so movies from 1990-1994 have little or no earlier history in the data
history_columns = ["dir_prior_films", "dir_prior_known_films", "dir_no_track_record",
                   "lead_actor_films_5yr", "company_films_5yr", "franchise_prior_films"]
period = pd.cut(movies["release_year"], bins=[1989, 1994, 1999, 2004, 2009, 2014, 2019, 2025],
                labels=["1990-94", "1995-99", "2000-04", "2005-09", "2010-14", "2015-19", "2020-25"])
movies.groupby(period, observed=True)[history_columns].mean().round(2)
```

### Director rating history from earlier movies
Link the broader matched movie sample to IMDb director IDs and pool each target movie's directors' strictly earlier-year movies. The raw prior mean remains available for inspection; select a smoothed mean, supporting count (`log1p`), and no-history flag for modeling.

The smoothed mean uses a five-film prior at the overall mean rating of strictly earlier reference movies; use the neutral rating-scale midpoint 5.0 when none exist. Give each earlier movie equal weight, counting co-directed titles once. Current/same-year/later movies are excluded. The reference includes matched movies without known finances, but its history still begins in 1990 and is not a full filmography.

These are retrospective rating snapshots, not historically archived ratings as of release. The features support rolling predictions; final snapshot ratings on earlier movies may differ from what was known at the time.

```python
# Ratings are observed TMDB snapshots linked to IMDb director IDs.
# Pool co-directors' earlier titles without double-counting the same movie.
def build_director_rating_history(targets, reference, prior_strength=5.0):
    """Earlier-year rating histories; never include the target, same-year, or later films."""
    if prior_strength <= 0:
        raise ValueError("prior_strength must be positive")
    reference = reference.copy()
    reference["release_year"] = pd.to_datetime(reference["release_date"]).dt.year
    reference = reference.loc[
        reference["vote_count"].gt(0)
        & reference["vote_average"].between(0, 10)
    ].dropna(subset=["nconst", "release_year", "tmdb_id"])
    # Movie-level ratings/dates must agree across director-pair rows.
    consistency = reference.groupby("tmdb_id")[["vote_average", "release_year"]].nunique()
    assert consistency.le(1).all().all()
    reference = reference.drop_duplicates(["tmdb_id", "nconst"])
    unique_films = reference.drop_duplicates("tmdb_id")
    prior_by_year = {}
    for year in targets["release_year"].unique():
        earlier_global = unique_films.loc[unique_films["release_year"] < year, "vote_average"]
        prior_by_year[year] = float(earlier_global.mean()) if len(earlier_global) else 5.0
    careers = {}
    for director, movie_id, year, rating in zip(
        reference["nconst"], reference["tmdb_id"], reference["release_year"], reference["vote_average"],
    ):
        careers.setdefault(director, []).append((movie_id, year, rating))
    histories = []
    for row in targets.itertuples():
        earlier = {}
        for director in str(row.imdb_director_ids).split("|"):
            for movie_id, year, rating in careers.get(director, []):
                if movie_id != row.tmdb_id and year < row.release_year:
                    earlier[movie_id] = rating
        count = len(earlier)
        total = sum(earlier.values())
        histories.append({
            "dir_prior_rating_mean": total / count if count else np.nan,
            "dir_prior_rated_films": count,
            "dir_prior_rating_smoothed": (
                total + prior_strength * prior_by_year[row.release_year]
            ) / (count + prior_strength),
            "dir_no_rating_history": int(count == 0),
        })
    return pd.DataFrame(histories, index=targets.index)

rating_reference = pd.read_csv(
    "../data/combined_dataset.csv",
    usecols=["tmdb_id", "nconst", "release_date", "vote_average", "vote_count"],
    dtype={"nconst": "string"}, parse_dates=["release_date"],
)
rating_history = build_director_rating_history(movies, rating_reference)
movies = movies.join(rating_history)
assert movies["dir_prior_rating_smoothed"].between(0, 10).all()
print("Movies with observed earlier director ratings:", int(movies["dir_prior_rated_films"].gt(0).sum()))
display(movies[["title_tmdb", "release_year", "dir_prior_rated_films", "dir_prior_rating_mean", "dir_prior_rating_smoothed"]].head(10))
```

### Step 11 Choose the model columns
The model table contains 14 numeric, 2 categorical, and 24 binary predictors, plus 4 identifier/date columns and two targets (46 columns total). `success` is the existing binary financial target; `vote_average` is the TMDB rating target for regression. Neither target nor the identifier/date columns belongs in the predictor matrix for either model. Log budget is included; raw budget, revenue/ratio, popularity, and vote count are excluded. Fit categorical encoding and numerical preprocessing on training data only. The three-class financial target remains to be defined separately.

### Revised predictors
Keep budget plots above and include `log_budget` as a numeric predictor. Add the 1985–1990 lead-actor supplement, smooth director success rates using a five-film prior, and apply `log1p` to experience counts. Scaling and imputation happen later using training data only.

History features use earlier years (franchises use earlier dates). Retrospective sources do not establish when final outcomes were actually reported. When evaluating a later multi-year block, these features represent rolling predictions with earlier outcomes available, rather than a single frozen forecast at the split date.

```python
# Smooth director outcomes using only earlier release years.
DIRECTOR_PRIOR_STRENGTH = 5.0
past_global_rate = movies["release_year"].map(
    {year: movies.loc[movies["release_year"] < year, "success"].mean()
     for year in movies["release_year"].unique()}
).fillna(0.5)
movies["dir_prior_success_rate_smoothed"] = (
    movies["dir_prior_success_rate"].fillna(0) * movies["dir_prior_known_films"]
    + DIRECTOR_PRIOR_STRENGTH * past_global_rate
) / (movies["dir_prior_known_films"] + DIRECTOR_PRIOR_STRENGTH)

# Require the supplement to describe this exact movie cohort and identities.
backfill = pd.read_csv("../data/lead_actor_backfill.csv", dtype={"imdb_id": "string"})
assert backfill["tmdb_id"].is_unique
assert set(backfill["tmdb_id"]) == set(movies["tmdb_id"])
expected_ids = movies.set_index("tmdb_id")["imdb_id"].sort_index()
pd.testing.assert_series_equal(
    backfill.set_index("tmdb_id")["imdb_id"].sort_index(), expected_ids,
    check_dtype=False,
)
movies = movies.merge(
    backfill[["tmdb_id", "imdb_extra_films_5yr", "backfill_unavailable"]],
    on="tmdb_id", how="left", validate="one_to_one", sort=False,
)
assert movies["imdb_extra_films_5yr"].ge(0).all()
movies["lead_actor_films_5yr_backfilled"] = (
    movies["lead_actor_films_5yr"] + movies["imdb_extra_films_5yr"]
)
EXPERIENCE_COUNTS = [
    "dir_prior_films", "dir_years_active", "dir_prior_known_films", "dir_prior_rated_films",
    "lead_actor_films_5yr_backfilled", "company_films_5yr", "franchise_prior_films",
]
for column in EXPERIENCE_COUNTS:
    assert movies[column].ge(0).all()
    movies[column + "_log1p"] = np.log1p(movies[column])

# IDs and dates support inspection/splitting; they are not predictors.
ID_COLUMNS = ["tmdb_id", "imdb_id", "title_tmdb", "release_date"]
TARGET = "success"  # Existing financial target; used in the exploratory correlations below.
RATING_TARGET = "vote_average"
TARGET_COLUMNS = [TARGET, RATING_TARGET]
NUMERIC_FEATURES = [
    "runtime", "log_budget", "release_year", "n_genres", "imdb_director_count",
    "dir_prior_success_rate_smoothed", "dir_prior_rating_smoothed",
] + [column + "_log1p" for column in EXPERIENCE_COUNTS]
CATEGORICAL_FEATURES = ["us_certification", "release_month"]
BINARY_FEATURES = [
    "is_sequel", "is_english", "dir_first_film", "dir_no_track_record", "backfill_unavailable", "dir_no_rating_history",
] + genre_columns
FEATURE_COLUMNS = NUMERIC_FEATURES + CATEGORICAL_FEATURES + BINARY_FEATURES
assert not set(TARGET_COLUMNS) & set(FEATURE_COLUMNS)
model_data = movies[ID_COLUMNS + FEATURE_COLUMNS + TARGET_COLUMNS].copy()
assert model_data["tmdb_id"].is_unique
assert not {"budget", "revenue", "revenue_to_budget", "vote_count", "popularity"} & set(model_data.columns)
print("Numeric:", len(NUMERIC_FEATURES), "| Categorical:", len(CATEGORICAL_FEATURES), "| Binary:", len(BINARY_FEATURES))
model_data.head()
assert model_data[RATING_TARGET].notna().all()
assert model_data[RATING_TARGET].between(0, 10).all()
# Training: X = model_data[FEATURE_COLUMNS]; choose one target column as y.
```

The heatmap describes numeric correlations with one another and with `success` in this cohort. Treat it as exploration, not evidence of predictive improvement. Any learned feature selection must happen inside training/validation.

```python
# Correlation between numeric features and the target
# Strongly correlated features carry overlapping information, which matters for logistic regression and PCA
corr = model_data[NUMERIC_FEATURES + [TARGET]].corr()

fig, ax = plt.subplots(figsize=(9, 7))
image = ax.imshow(corr, cmap="RdBu_r", vmin=-1, vmax=1)
ax.set_xticks(range(len(corr.columns)))
ax.set_xticklabels(corr.columns, rotation=60, ha="right", fontsize=8)
ax.set_yticks(range(len(corr.columns)))
ax.set_yticklabels(corr.columns, fontsize=8)
for i in range(len(corr.columns)):
    for j in range(len(corr.columns)):
        value = corr.iloc[i, j]
        text_color = "white" if abs(value) > 0.6 else "black"     # readable on dark cells
        ax.text(j, i, f"{value:.2f}", ha="center", va="center", fontsize=6, color=text_color)
fig.colorbar(image, ax=ax, label="Correlation")
ax.set_title("Correlation of numeric features and the target")
plt.tight_layout()
plt.show()

corr[TARGET].drop(TARGET).sort_values().round(3)
```

### Step 14 Build the broader rating dataset

Use the same director careers, strictly earlier rating histories, and financially eligible reference for prior financial success—not an invented current financial label. Keep the existing actor supplement for known matches; newly included early-window movies without a supplement are explicitly flagged, rather than treating absent matches as complete history. Source, combined, and cleaned CSVs are read-only inputs.

The success table retains its original cohort/values under the new `model_data_success.csv` name. Both exports stop rather than replace a differing existing snapshot.

Use `log10(budget)` for positive reported budgets. Unknown/nonpositive budgets remain missing, are imputed inside each training pipeline, and do not remove rating movies.

```python
# Rating eligibility is independent of budget and revenue.
combined_rating = pd.read_csv(
    "../data/combined_dataset.csv",
    dtype={"imdb_id": "string", "nconst": "string", "collection_id": "Int64", "startYear": "Int64"},
    parse_dates=["release_date"],
)
director_lists = combined_rating.groupby("tmdb_id", as_index=False).agg(
    imdb_director_ids=("nconst", lambda values: "|".join(values.dropna().drop_duplicates())),
    imdb_director_count=("nconst", "nunique"),
)
rating_movies = (
    combined_rating.drop(columns=["nconst", "director_imdb"])
    .drop_duplicates("tmdb_id")
    .merge(director_lists, on="tmdb_id", validate="one_to_one", sort=False)
)
rating_movies = rating_movies.loc[
    rating_movies["release_date"].notna()
    & rating_movies["vote_average"].between(0, 10)
    & rating_movies["vote_count"].gt(0)
].copy()
# Unknown/nonpositive budgets stay missing; models impute using training data only.
rating_movies["log_budget"] = np.log10(rating_movies["budget"].where(rating_movies["budget"].gt(0)))
rating_movies["release_year"] = rating_movies["release_date"].dt.year
rating_movies["release_month"] = rating_movies["release_date"].dt.month
# Current source years are complete; do not invent IMDb history ordering if they become missing.
assert rating_movies["startYear"].notna().all()
rating_movies.loc[rating_movies["runtime"].le(0), "runtime"] = np.nan
rating_movies["us_certification"] = rating_movies["us_certification"].fillna("Unknown").replace({"R": "R/NC-17", "NC-17": "R/NC-17"})
for column in ["top_cast", "production_companies"]:
    rating_movies[column] = rating_movies[column].fillna("")
for genre, column in zip(genre_list, genre_columns):
    rating_movies[column] = rating_movies["genres"].fillna("").str.split("|").map(lambda names: int(genre in names))
rating_movies["n_genres"] = rating_movies["genres"].fillna("").str.split("|").map(lambda names: len([name for name in names if name]))
rating_movies["is_english"] = rating_movies["original_language"].eq("en").astype(int)

# Reuse the same IMDb careers and known-finance reference built above.
career_values = [director_history_imdb(ids, year) for ids, year in zip(rating_movies["imdb_director_ids"], rating_movies["startYear"])]
rating_movies["dir_prior_films"] = [value[0] for value in career_values]
rating_movies["dir_years_active"] = [value[1] for value in career_values]
rating_movies["dir_first_film"] = rating_movies["dir_prior_films"].eq(0).astype(int)
financial_values = [director_track_record(ids, year) for ids, year in zip(rating_movies["imdb_director_ids"], rating_movies["release_year"])]
rating_movies["dir_prior_known_films"] = [value[0] for value in financial_values]
rating_movies["dir_prior_success_rate"] = [value[1] for value in financial_values]
rating_movies["dir_no_track_record"] = rating_movies["dir_prior_known_films"].eq(0).astype(int)
financial_prior = rating_movies["release_year"].map({
    year: movies.loc[movies["release_year"] < year, "success"].mean()
    for year in rating_movies["release_year"].unique()
}).fillna(0.5)
rating_movies["dir_prior_success_rate_smoothed"] = (
    rating_movies["dir_prior_success_rate"].fillna(0) * rating_movies["dir_prior_known_films"]
    + DIRECTOR_PRIOR_STRENGTH * financial_prior
) / (rating_movies["dir_prior_known_films"] + DIRECTOR_PRIOR_STRENGTH)
rating_movies = rating_movies.join(build_director_rating_history(rating_movies, rating_reference))

rating_movies["lead_actor_films_5yr"] = [
    sum(year - HISTORY_YEARS <= earlier < year for earlier in actor_years.get(cast.split("|")[0], []))
    for cast, year in zip(rating_movies["top_cast"], rating_movies["release_year"])
]
rating_movies["company_films_5yr"] = [
    max((sum(year - HISTORY_YEARS <= earlier < year for earlier in company_years.get(company, [])) for company in companies.split("|") if company), default=0)
    for companies, year in zip(rating_movies["production_companies"], rating_movies["release_year"])
]
rating_movies["franchise_prior_films"] = [
    sum(earlier < date for earlier in collection_dates.get(collection, [])) if pd.notna(collection) else 0
    for collection, date in zip(rating_movies["collection_id"], rating_movies["release_date"])
]
rating_movies["is_sequel"] = rating_movies["franchise_prior_films"].gt(0).astype(int)

# Existing actor supplement matches only the original success cohort. Do not fabricate new matches.
rating_movies = rating_movies.merge(
    backfill[["tmdb_id", "imdb_id", "imdb_extra_films_5yr", "backfill_unavailable"]],
    on=["tmdb_id", "imdb_id"], how="left", validate="one_to_one", sort=False,
)
needs_early_backfill = rating_movies["release_year"].between(1990, 1995)
rating_movies["backfill_unavailable"] = rating_movies["backfill_unavailable"].fillna(needs_early_backfill.astype(int)).astype(int)
rating_movies["lead_actor_films_5yr_backfilled"] = rating_movies["lead_actor_films_5yr"] + rating_movies["imdb_extra_films_5yr"].fillna(0)
for column in EXPERIENCE_COUNTS:
    assert rating_movies[column].ge(0).all()
    rating_movies[column + "_log1p"] = np.log1p(rating_movies[column])
rating_movies = rating_movies.sort_values("release_date", kind="stable").reset_index(drop=True)
rating_model_data = rating_movies[ID_COLUMNS + FEATURE_COLUMNS + [RATING_TARGET]].copy()
assert rating_model_data["tmdb_id"].is_unique
assert rating_model_data[RATING_TARGET].notna().all()
assert not {"success", "budget", "revenue", "revenue_to_budget", "vote_count", "popularity"} & set(rating_model_data)

# Check shared movies retain exactly the success table's predictors and rating values.
shared_columns = [column for column in model_data if column != "success"]
shared_rating = rating_model_data.set_index("tmdb_id").reindex(model_data["tmdb_id"])
shared_success = model_data.set_index("tmdb_id")
pd.testing.assert_frame_equal(shared_rating[shared_success.columns.drop("success")], shared_success.drop(columns="success"), check_dtype=False)

rating_output = Path("../data/model_data_rating.csv")
rating_serialized = rating_model_data.to_csv(index=False)
if rating_output.exists():
    pd.testing.assert_frame_equal(pd.read_csv(rating_output), pd.read_csv(io.StringIO(rating_serialized)), check_dtype=False)
    print("Existing rating snapshot matches; no CSV was changed.")
else:
    with rating_output.open("x") as handle:
        handle.write(rating_serialized)
print(f"Rating table: {len(rating_model_data):,} movies, {len(rating_model_data.columns)} columns")
print(f"Success table: {len(model_data):,} movies, {len(model_data.columns)} columns")
print("Rating rows without reported positive budget or revenue:", int((rating_movies["budget"].le(0) | rating_movies["revenue"].le(0)).sum()))
print("Rating rows with unavailable early actor supplement:", int(rating_model_data["backfill_unavailable"].sum()))
```
