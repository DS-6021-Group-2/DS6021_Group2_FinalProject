```python
#packages 
import pandas as pd
```

```python
# load in datasets
dbm = pd.read_csv("data/dmb_movies.csv")
tmdb = pd.read_csv("data/tmdb_movies.csv")
```

```python
# Check imdb_id for Missing Values and Duplicates

print(tmdb["imdb_id"].isna().sum(), dmb["imdb_id"].isna().sum())   # missing ids
print(tmdb["imdb_id"].duplicated().sum(), dmb["imdb_id"].duplicated().sum())  # duplicates
```

Each table has 10 rows with a missing `imdb_id` and 9 duplicates. The 9 "duplicates" in each are most likely just  blanks being counted as repeats of each other

```python
# Drop missing rows and double check (00 on output is good)
tmdb = tmdb.dropna(subset=["imdb_id"])
dmb = dmb.dropna(subset=["imdb_id"])

print(tmdb["imdb_id"].duplicated().sum(), dmb["imdb_id"].duplicated().sum())
```

```python
#merge
combined = tmdb.merge(dmb, on="imdb_id", how="inner", suffixes=("_tmdb", "_dmb"))
print(tmdb.shape, dmb.shape, combined.shape)
combined.head()
```

```python
# Inspect Combined Column Names
print(combined.columns.tolist())
```

```python
# Compare Titles Across the Two Sources
print((combined["title_tmdb"] == combined["title_dmb"]).mean())
```

having a result of 1.0 means the titles were successfully merged

```python
## Save the Combined Dataset to CSV
combined.to_csv("combined_dataset.csv", index=False)
```
